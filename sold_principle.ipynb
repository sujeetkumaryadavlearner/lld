# S -Single Responsibility Principle

```python
class Waiter:
    def __init__(self):
        self.name="Waiter"
    def server(self):
        print("pick and serving")

class Captian:
    def __init__(self):
        self.name="Captain"
    def taker_order(self):
        print("Taking order")

class Greeter:
    def __init__(self):
        self.name="Greeter"
    def greet(self):
        print("Welcome to our restaurant")

class chef:
    def __init__(self):
        self.name="Chef"
    def cook(self):
        print("Cooking food")

class Restaurant:
    def __init__(self):
        self.waiter=Waiter()
        self.captain=Captian()
        self.greeter=Greeter()
        self.chef=chef()

    def run(self):
        self.greeter.greet()
        self.captain.taker_order()
        self.chef.cook()
        self.waiter.server()
```

```python
r1=Restaurant()
r1.run()
```

# O -Open Closed Principle

```python
from abc import ABC,abstractmethod

class Payment_Method(ABC):

    @abstractmethod
    def pay(self):
        pass
    
class UPI_Payment(Payment_Method):
    def pay(self):
        print("Paying through UPI")

class Cash_Payment(Payment_Method):
    def pay(self):
        print("Paying through Cash")

class Card_Payment(Payment_Method):
    def pay(self):
        print("Paying through Card")
        
class Payment:
    def __init__(self,method):
        self.method=method

    def make_payment(self):
        self.method.pay()
        
```

```python
Payment1=Payment(UPI_Payment())
Payment1.make_payment()
Payment2=Payment(Cash_Payment())
Payment2.make_payment()
```

# Liskov Substitution Principle

```python
class booking:
    
    def make_booking(self,count):
        print(f"Booking for {count} people is done")

class online_booking(booking):
    def make_booking(self,count):
        print(f"Online booking for {count} people is done")


class counter_booking(booking):
    def make_booking(self,count):
        print(f"Counter booking for {count} people is done")


class Bookticker:
    def __init__(self,method):
        self.method=method

    def book_ticket(self,count):
        self.method.make_booking(count)


booking1=Bookticker(online_booking())
booking1.book_ticket(4)
booking2=Bookticker(counter_booking())
booking2.book_ticket(2)
```
