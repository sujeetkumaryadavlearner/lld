🧩 Design Question — Notification System

Design a system where a user can receive notifications through:

Email

SMS

WhatsApp

The system should allow adding new notification types in the future without modifying existing code.

```python
from abc import ABC,abstractmethod
```

```python
class Notification(ABC):
    
    @abstractmethod

    def send_notification(self,content):
        pass

class email_notification(Notification):
    
    def send_notification(self, content):
        print("notification send using email",{content})

class sms_notification(Notification):
    
    def send_notification(self, content):
        print("notification send using sms",{content})

    
class whatsapp_notification(Notification):
    
    def send_notification(self, content):
        print("notification send using whaapp",{content})



class Notification_sender:
    
    def __init__(self,method):
        self.method=method
    
    def send_notification(self):
        self.method.send_notification("hello")


```

```python
ns1=Notification_sender(whatsapp_notification())
ns2=Notification_sender(email_notification())
ns3=Notification_sender(sms_notification())
```
